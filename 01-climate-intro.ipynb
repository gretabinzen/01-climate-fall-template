**Impacts of Climate Change on the San Juan Mountains in Southwest Colorado.** 

*  There is **limited research** pertaining directly to the San Juan Mountains so most knowledge comes from studies of similar areas (Mountain Studies, 2014).
* **Warming temperatues.** Between 1977 and 2007 temperatures in Southwest Colorado rose 2 degrees (Mountain Studies, 2014). In August 2026, 25 days of the month had temperature highs that exceeded the historic normal temperature range and 10 days with record highs (NOAA, 2026). 
* **Declining snowpack.** Areas of Colorado have seen up to an 80% decrease in snowpack over the last 70 years. Snow is the largest source of water for the state so decreasing snowpack impacts our forests and agricultural systems, increases wildfire, and also heavily impacts industries such as rafting and skiing (Bonnell, 2024).
* **Wildfires are burning hotter and faster.** Western US has always experienced wildfires but due to years of fire supression and increased drought, fires today are bigger and hotter than ever before. This impacts the ability of the forest to regrow, creates a hydrophobic layer of soil, and impacts wildlife in the area among other impacts (Coop, 2026). 
* **Increased algae in alpine lakes.** Warming temperatures combined with air pollution is impacting alpine lakes throughout Colorado (Allen, 2025).
